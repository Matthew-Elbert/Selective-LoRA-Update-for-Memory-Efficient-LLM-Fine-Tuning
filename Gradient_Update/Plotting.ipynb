```python
import json
import matplotlib.pyplot as plt
```

50%

```python
with open('results_50%_20260222_103201/gpu_usage_log.json') as f:
    gpu_50=json.load(f)

with open('results_50%_20260222_103201/ram_usage_log.json') as f:
    ram_50=json.load(f)
    
with open('results_50%_20260222_103201/log.json') as f:
    log_50=json.load(f)
```

```python
#Plot Loss

loss=[x['loss'] for x in log_50[:-2]]
step=[x['step'] for x in log_50[:-2]]

plt.plot(step,loss)
plt.xlabel('Number of Steps')
plt.ylabel('Loss')
plt.title('Loss vs Number of Steps (50% Components)')
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Extract GPU and RAM usage data
# gpu_usage = gpu
# ram_usage = ram
# time_stamps = [entry['timestamp'] for entry in gpu]  # Assuming both logs have the same timestamps

chunk_size = 50
gpu_usage = [
    gpu_50[i]
    for i in range(0, len(gpu_50), chunk_size)
]

ram_usage = [
    ram_50[i]
    for i in range(0, len(ram_50), chunk_size)
]


time_stamps = [x*5 for x in range(len(gpu_usage))]  # Using index as timestamp if actual timestamps are not available
# time_deltas = (time_stamps - time_stamps[0]).total_seconds() / 60  # Convert to minutes
# Plot GPU and RAM usage over time
plt.figure(figsize=(12, 6))
plt.plot(time_stamps, gpu_usage, label='GPU Usage ', color='blue')
plt.plot(time_stamps, ram_usage, label='RAM Usage', color='orange')
plt.xlabel('Time')
plt.ylabel('Usage')
plt.title('GPU and RAM Usage Over Time (50% Components)')
plt.legend()
plt.grid(True)
plt.show()
```
