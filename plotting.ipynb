```python
import json

with open('./gpu_usage_log.json', 'r') as file:
    gpu = json.load(file)
```

```python
import json

with open('./ram_usage_log.json', 'r') as file:
    ram = json.load(file)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Extract GPU and RAM usage data
# gpu_usage = gpu
# ram_usage = ram
# time_stamps = [entry['timestamp'] for entry in gpu]  # Assuming both logs have the same timestamps

chunk_size = 50
gpu_usage = [
    gpu[i]
    for i in range(0, len(gpu), chunk_size)
]

ram_usage = [
    ram[i]
    for i in range(0, len(ram), chunk_size)
]


time_stamps = [x*5 for x in range(len(gpu_usage))]  # Using index as timestamp if actual timestamps are not available
# time_deltas = (time_stamps - time_stamps[0]).total_seconds() / 60  # Convert to minutes
# Plot GPU and RAM usage over time
plt.figure(figsize=(12, 6))
plt.plot(time_stamps, gpu_usage, label='GPU Usage ', color='blue')
plt.plot(time_stamps, ram_usage, label='RAM Usage', color='orange')
plt.xlabel('Time')
plt.ylabel('Usage')
plt.title('GPU and RAM Usage Over Time')
plt.legend()
plt.grid(True)
plt.show()
```
